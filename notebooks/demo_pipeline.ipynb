# VLN Live Perception Pipeline — Demo Notebook

Combines:
- **AP-VLM** (Sripada et al., 2024): Active Perception via Vision-Language Models
- **Cola** (Chen et al., NeurIPS 2023): LLM coordinates dual VLMs (BLIP + YOLO)

This notebook runs the pipeline on a **video file** (no webcam/robot needed for demo).
Swap `source: webcam` in config.yaml to run on live camera.

```python
import sys, os
sys.path.insert(0, '..')
from main import run, load_config

# Override config for notebook demo
cfg = {
    'input': {
        'source': 'video_file',       # change to 'webcam' for live camera
        'video_path': 'demo.mp4',     # path to your video
    },
    'perception': {
        'device': 'cpu',
        'blip_model': 'Salesforce/blip-image-captioning-base',
        'yolo_model': 'yolov8n.pt',
        'midas_variant': 'DPT_Hybrid',
        'yolo_conf_threshold': 0.25,
        'compute_embedding': False,
    },
    'coordinator': {
        'backend': 'hf',
        'hf_model': 'google/flan-t5-base',
        'openai_model': 'gpt-4o-mini',
        'openai_api_key': '',
    },
    'active_perception': {
        'max_iterations': 5,
        'confidence_threshold': 0.6,
    },
    'navigation': {
        'instruction': 'Navigate to the exit door and avoid obstacles.',
    },
    'pipeline': {
        'frame_interval': 10,
        'max_frames': 20,
        'show_preview': False,
    },
    'output': {
        'results_dir': '../results',
        'figures_dir': '../figures',
        'session_name': 'notebook_demo',
    },
}
```

```python
# Run pipeline
output_paths = run(cfg, use_coordinator=True)
print('\nOutput files:')
for k, v in output_paths.items():
    print(f'  {k}: {v}')
```
